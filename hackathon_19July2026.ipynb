## Data Preprocessing

```python
print('Missing values in df_train:')
print(df_train.isnull().sum() * 100 / len(df_train))

print('\nMissing values in df_test:')
print(df_test.isnull().sum() * 100 / len(df_test))
```

## Handle Missing Values

```python
# Drop rows in df_train where 'Attrition' or 'EmployeeID' are null
df_train.dropna(subset=['Attrition', 'EmployeeID'], inplace=True)

print("Shape of df_train after dropping rows with missing Attrition/EmployeeID:", df_train.shape)

# Separate numerical and categorical columns
numerical_cols_train = df_train.select_dtypes(include=['float64', 'int64']).columns
categorical_cols_train = df_train.select_dtypes(include=['object']).columns

numerical_cols_test = df_test.select_dtypes(include=['float64', 'int64']).columns
categorical_cols_test = df_test.select_dtypes(include=['object']).columns

# Impute missing numerical values with the median
for col in numerical_cols_train:
    if df_train[col].isnull().any():
        median_val = df_train[col].median()
        df_train[col].fillna(median_val, inplace=True)

for col in numerical_cols_test:
    if df_test[col].isnull().any():
        # Use median from df_train for consistency if applicable, otherwise from df_test
        if col in df_train.columns: # Check if column exists in train for consistent imputation
            median_val = df_train[col].median()
        else:
            median_val = df_test[col].median()
        df_test[col].fillna(median_val, inplace=True)

# Impute missing categorical values with the mode
for col in categorical_cols_train:
    if df_train[col].isnull().any():
        mode_val = df_train[col].mode()[0]
        df_train[col].fillna(mode_val, inplace=True)

for col in categorical_cols_test:
    if df_test[col].isnull().any():
        # Use mode from df_train for consistency if applicable, otherwise from df_test
        if col in df_train.columns: # Check if column exists in train for consistent imputation
            mode_val = df_train[col].mode()[0]
        else:
            mode_val = df_test[col].mode()[0]
        df_test[col].fillna(mode_val, inplace=True)

print('\nMissing values in df_train after imputation:')
print(df_train.isnull().sum() * 100 / len(df_train))

print('\nMissing values in df_test after imputation:')
print(df_test.isnull().sum() * 100 / len(df_test))
```

## Prepare Data for Model Training

```python
from sklearn.model_selection import train_test_split

# Convert 'Attrition' to int type (0 or 1)
df_train['Attrition'] = df_train['Attrition'].astype(int)

# Identify categorical columns for one-hot encoding
categorical_cols = df_train.select_dtypes(include=['object']).columns

# Apply one-hot encoding to training data
df_train_encoded = pd.get_dummies(df_train, columns=categorical_cols, drop_first=True)

# Apply one-hot encoding to test data
df_test_encoded = pd.get_dummies(df_test, columns=categorical_cols, drop_first=True)

# Align columns - very important for consistent training and testing sets
train_cols = set(df_train_encoded.columns)
test_cols = set(df_test_encoded.columns)

missing_in_test = list(train_cols - test_cols)
for col in missing_in_test:
    if col != 'Attrition': # 'Attrition' is the target, should not be in test features
        df_test_encoded[col] = 0 # Add missing columns to test set and fill with 0

missing_in_train = list(test_cols - train_cols)
for col in missing_in_train:
    df_train_encoded[col] = 0 # Add missing columns to train set and fill with 0

# Ensure the order of columns is the same, excluding 'Attrition' for test set
common_cols = list(set(df_train_encoded.columns) & set(df_test_encoded.columns))

# Separate features (X) and target (y) for the full training dataset
X_full_train = df_train_encoded.drop(columns=['Attrition', 'EmployeeID'])
y_full_train = df_train_encoded['Attrition']

# Split the full training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_full_train, y_full_train, test_size=0.1, random_state=42, stratify=y_full_train)

# Prepare test features
X_test = df_test_encoded.drop(columns=['EmployeeID'])

# Align columns for X_test with X_train after dropping Attrition
X_test = X_test[X_train.columns]

print("Shape of X_train (for model training):", X_train.shape)
print("Shape of y_train (for model training):", y_train.shape)
print("Shape of X_val (for validation):", X_val.shape)
print("Shape of y_val (for validation):", y_val.shape)
print("Shape of X_test (for final prediction):", X_test.shape)
print("Columns in X_train match X_test in order and presence:", list(X_train.columns) == list(X_test.columns))
```

```python
# Re-running initial Random Forest training and prediction with new split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Initialize the Random Forest Classifier
model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model on the split training set
model.fit(X_train, y_train)

# Predict attrition on the validation set for accuracy calculation
y_val_pred = model.predict(X_val)

# Calculate accuracy on the validation set
val_accuracy = accuracy_score(y_val, y_val_pred)

print(f"Accuracy on the validation set (RandomForest): {val_accuracy:.4f}")
```

```python
# Re-running Logistic Regression with new split
from sklearn.linear_model import LogisticRegression

log_reg_model = LogisticRegression(random_state=42, solver='liblinear', max_iter=1000)
log_reg_model.fit(X_train, y_train)
y_val_pred_lr = log_reg_model.predict(X_val)
accuracy_lr = accuracy_score(y_val, y_val_pred_lr)
print(f"Logistic Regression Validation Accuracy: {accuracy_lr:.4f}")
```

```python
# Re-running Support Vector Machine with new split
from sklearn.svm import SVC

svm_model = SVC(random_state=42, kernel='linear', C=1.0, probability=True)
svm_model.fit(X_train, y_train)
y_val_pred_svm = svm_model.predict(X_val)
accuracy_svm = accuracy_score(y_val, y_val_pred_svm)
print(f"Support Vector Machine Validation Accuracy: {accuracy_svm:.4f}")
```

```python
# Re-running Gradient Boosting Classifier with new split
from sklearn.ensemble import GradientBoostingClassifier

gbc_model = GradientBoostingClassifier(random_state=42, n_estimators=100, learning_rate=0.1)
gbc_model.fit(X_train, y_train)
y_val_pred_gbc = gbc_model.predict(X_val)
accuracy_gbc = accuracy_score(y_val, y_val_pred_gbc)
print(f"Gradient Boosting Classifier Validation Accuracy: {accuracy_gbc:.4f}")
```

```python
# Re-running Model Comparison and Selection with new split
# Original Random Forest model accuracy (now re-calculated)
rf_accuracy = val_accuracy

model_accuracies = {
    'RandomForestClassifier': rf_accuracy,
    'LogisticRegression': accuracy_lr,
    'SVC': accuracy_svm,
    'GradientBoostingClassifier': accuracy_gbc
}

best_model_name = max(model_accuracies, key=model_accuracies.get)
best_accuracy = model_accuracies[best_model_name]

print("\n--- Model Comparison (Updated) ---")
for model, accuracy in model_accuracies.items():
    print(f"{model}: {accuracy:.4f}")

print(f"\nBest model for hyperparameter tuning is: {best_model_name} with accuracy: {best_accuracy:.4f}")

# Assign the best model for tuning
if best_model_name == 'RandomForestClassifier':
    model_for_tuning = RandomForestClassifier(random_state=42)
elif best_model_name == 'LogisticRegression':
    model_for_tuning = LogisticRegression(random_state=42, solver='liblinear', max_iter=1000)
elif best_model_name == 'SVC':
    model_for_tuning = SVC(random_state=42, probability=True)
elif best_model_name == 'GradientBoostingClassifier':
    model_for_tuning = GradientBoostingClassifier(random_state=42)

print(f"\nProceeding with hyperparameter tuning for {best_model_name}.")
```

```python
# Re-running Hyperparameter Tuning with new split
from sklearn.model_selection import GridSearchCV

param_grids = {
    'RandomForestClassifier': {
        'n_estimators': [50, 100, 200],
        'max_depth': [None, 10, 20],
        'min_samples_split': [2, 5]
    },
    'LogisticRegression': {
        'C': [0.1, 1.0, 10.0],
        'penalty': ['l1', 'l2']
    },
    'SVC': {
        'C': [0.1, 1.0, 10.0],
        'kernel': ['linear', 'rbf']
    },
    'GradientBoostingClassifier': {
        'n_estimators': [50, 100, 200],
        'learning_rate': [0.01, 0.1, 0.2],
        'max_depth': [3, 5]
    }
}

param_grid = param_grids[best_model_name]

grid_search = GridSearchCV(estimator=model_for_tuning, param_grid=param_grid, cv=3, scoring='accuracy', n_jobs=-1, verbose=1)
grid_search.fit(X_train, y_train)

print(f"\nBest parameters for {best_model_name}: {grid_search.best_params_}")
print(f"Best cross-validation accuracy for {best_model_name}: {grid_search.best_score_:.4f}")

best_tuned_model = grid_search.best_estimator_
y_val_pred_tuned = best_tuned_model.predict(X_val)
tuned_accuracy = accuracy_score(y_val, y_val_pred_tuned)
print(f"Tuned {best_model_name} Validation Accuracy: {tuned_accuracy:.4f}")
```

```python
# Re-running Final Prediction and Submission with Tuned Model with new split
import pandas as pd

final_predictions = best_tuned_model.predict(X_test)
submission_df = pd.DataFrame({'EmployeeID': df_test['EmployeeID'], 'Attrition': final_predictions})

print(submission_df.head())
submission_df.to_csv('/content/drive/MyDrive/attrition_predictions_tuned.csv', index=False)

print("Predictions from the tuned model (80: 20 split) saved to 'attrition_predictions_tuned_80_20_split.csv'")
```

## Ways to Increase Random Forest Accuracy

Random Forest models are powerful, but their accuracy can often be further improved through several techniques:

1.  **Hyperparameter Tuning (which you've already done!)**:
    *   `n_estimators`: Increasing the number of trees in the forest can improve stability and reduce variance, but also increases computation time. You already tuned this.
    *   `max_depth`: Limiting the maximum depth of each tree helps prevent overfitting. You already tuned this.
    *   `min_samples_split` / `min_samples_leaf`: These control the minimum number of samples required to split an internal node or be at a leaf node, respectively. Increasing these values can prevent overfitting. You already tuned `min_samples_split`.
    *   `max_features`: The number of features to consider when looking for the best split. A smaller value reduces variance but increases bias. Tuning this can help.

2.  **Feature Engineering and Selection**:
    *   **Create new features**: Sometimes, combining existing features or creating new ones (e.g., ratios, interactions) can provide the model with more meaningful information.
    *   **Feature scaling**: While tree-based models like Random Forest are not sensitive to feature scaling, it's good practice for other models and can sometimes indirectly help by making other preprocessing steps more effective.
    *   **Feature selection**: Removing irrelevant or redundant features can reduce noise and potentially improve model performance and speed. Techniques include RFE (Recursive Feature Elimination), SelectKBest, or using feature importances from an initial model run.

3.  **Handling Imbalanced Datasets**:
    *   If one class significantly outnumbers the other (as might be the case with attrition), the model might be biased towards the majority class.
    *   **Techniques**: Oversampling the minority class (e.g., SMOTE), undersampling the majority class, or using class weights in the model (e.g., `class_weight='balanced'` in scikit-learn's Random Forest).

4.  **Cross-Validation Strategy**:
    *   Ensure your cross-validation strategy is robust (e.g., `StratifiedKFold` for imbalanced classification tasks) to get a reliable estimate of model performance and prevent overfitting to a specific train/validation split.

5.  **Ensemble Methods beyond Random Forest**:
    *   While Random Forest is an ensemble method (Bagging), you could consider combining it with other types of models or exploring boosting techniques (like XGBoost, LightGBM, CatBoost) that often achieve higher accuracy, albeit with increased complexity.

## Stacking Ensemble for Increased Accuracy

Stacking (Stacked Generalization) is an ensemble machine learning method that learns to combine the predictions from multiple base models. The base models are trained on the complete training dataset, and then a meta-model (or meta-learner) is trained on the predictions of these base models. This approach can often achieve higher accuracy than any individual model by leveraging the strengths of different algorithms.

Here, we'll implement a stacking ensemble using the `RandomForestClassifier`, `LogisticRegression`, `SVC`, and `GradientBoostingClassifier` as base estimators, and a `LogisticRegression` as the meta-learner.

```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression

# Define base models (estimators). Use the best tuned Random Forest.
estimators = [
    ('rf', best_tuned_model), # Use the best tuned Random Forest
    ('lr', LogisticRegression(random_state=42, solver='liblinear', max_iter=1000)),
    ('svc', SVC(random_state=42, kernel='linear', C=1.0, probability=True)),
    ('gbc', GradientBoostingClassifier(random_state=42, n_estimators=100, learning_rate=0.1))
]

# Define the meta-learner
# A simple Logistic Regression is often a good choice for the meta-learner
final_estimator = LogisticRegression(random_state=42, solver='liblinear', max_iter=1000)

# Create the Stacking Classifier
stacking_model = StackingClassifier(
    estimators=estimators,
    final_estimator=final_estimator,
    cv=3, # Use cross-validation to generate out-of-fold predictions for the meta-learner
    n_jobs=-1, # Use all available cores
    verbose=1
)

print("Training Stacking Classifier...")
# Train the stacking model on the training data
stacking_model.fit(X_train, y_train)
print("Stacking Classifier training complete.")

# Predict on the validation set using the stacking model
y_val_pred_stacking = stacking_model.predict(X_val)

# Calculate accuracy on the validation set
stacking_accuracy = accuracy_score(y_val, y_val_pred_stacking)

print(f"\nStacking Classifier Validation Accuracy: {stacking_accuracy:.4f}")
```

```python
# Make final predictions on the test set using the trained stacking model
final_predictions_stacking = stacking_model.predict(X_test)

# Create a DataFrame with EmployeeID and predictions
submission_df_stacking = pd.DataFrame({'EmployeeID': df_test['EmployeeID'], 'Attrition': final_predictions_stacking})

# Display the first few final predictions
print("\nStacking Model Predictions (first 5 rows):")
display(submission_df_stacking.head())

# Save the submission_df to a CSV file
submission_df_stacking.to_csv('attrition_predictions_stacking.csv', index=False)

print("Predictions from the stacking model saved to 'attrition_predictions_stacking.csv'")
```
